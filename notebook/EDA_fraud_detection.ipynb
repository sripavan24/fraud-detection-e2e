# under _Sampling

Bulid a sample dataset contain similiar distribution of normal transaction and fradlaet tranction


num of fraudulant transaction is                          ------>492

```python
legit_sample_data=legit.sample(n=492)
```

```python
sample_data_set=pd.concat([legit_sample_data,fraud],axis=0)
```

```python
sample_data_set.head()
```

```python
sample_data_set.describe()
```

```python
sample_data_set["Class"].value_counts()
```

```python
sample_data_set.groupby('Class').mean()
```

```python
sns.countplot(data=sample_data_set,x="Class",hue='Class')
plt.show()
```

```python
data1=sample_data_set.groupby("Class")['Amount'].mean()
```

```python
plt.figure(figsize=(6,6))
plt.pie(data1.values,
        labels=data1.index,
        autopct='%1.1f%%',
        startangle=90)
plt.title("Amount Distribution")
plt.show()


```

```python
plt.figure(figsize=(9,9))
sns.heatmap(sample_data_set.corr())
plt.show()
```

```python
sns.boxplot(x='Class', y='Amount', data=sample_data_set)
plt.show()
```

```python
sns.histplot(data=sample_data_set, x='Amount', hue='Class', bins=60)
plt.show()
```

```python
#plot the histogram of a variable from the dataset to see the skewness
import math
plt.figure(figsize=(20,20))
cols = 3
rows = math.ceil(len(df.columns) / cols)
for n, col in enumerate(df.columns[:-1]):
    plt.subplot(rows,3,n+1)
    sns.histplot(data=df, x=col, hue="Class", kde=True)
    plt.title(col)

plt.tight_layout()
plt.show()
```
